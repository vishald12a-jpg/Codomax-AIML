```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Student": [
        "Arun", "Bala", "Charan", "Divya", "Elena",
        "Farhan", "Gokul", "Hari", "Isha", "Jaya"
    ],

    "Maths": [85, 72, 90, 65, 78, 88, 56, 92, 74, 81],

    "Python": [90, 68, 95, 70, 82, 91, 60, 94, 76, 85],

    "DBMS": [78, 75, 88, 62, 80, 86, 58, 90, 72, 79],

    "AI_ML": [92, 70, 96, 68, 85, 89, 55, 95, 78, 87]
}

df = pd.DataFrame(data)

df
```

```python
print("First 5 rows:")
display(df.head())

print("Dataset shape:")
print(df.shape)

print("Column names:")
print(df.columns)

print("Dataset information:")
df.info()
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
df.loc[3, "Python"] = np.nan

df
print(df.isnull().sum())
```

```python
python_mean = df["Python"].mean()

print("Average Python mark:", python_mean)

df["Python"] = df["Python"].fillna(python_mean)

print(df.isnull().sum())
```

```python
print("Statistical Summary:")
display(df.describe())

subject_columns = ["Maths", "Python", "DBMS", "AI_ML"]

subject_averages = df[subject_columns].mean()

print("Average marks:")
print(subject_averages)

df["Total"] = df[subject_columns].sum(axis=1)

df["Average"] = df[subject_columns].mean(axis=1)

df[["Student", "Total", "Average"]]
```

```python
plt.figure(figsize=(8, 5))

plt.bar(subject_averages.index, subject_averages.values)

plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    df["Student"],
    df["Average"],
    marker="o"
)

plt.title("Student Average Marks")
plt.xlabel("Student")
plt.ylabel("Average Marks")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```

```python
def calculate_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


df["Grade"] = df["Average"].apply(calculate_grade)

df[["Student", "Average", "Grade"]]
grade_counts = df["Grade"].value_counts()

print(grade_counts)
plt.figure(figsize=(7, 7))

plt.pie(
    grade_counts.values,
    labels=grade_counts.index,
    autopct="%1.1f%%"
)

plt.title("Grade Distribution")

plt.show()
```

```python
print("===== FINAL ANALYSIS =====")

highest_subject = subject_averages.idxmax()
lowest_subject = subject_averages.idxmin()

highest_student = df.loc[df["Average"].idxmax(), "Student"]
highest_average = df["Average"].max()

print("Highest average subject:", highest_subject)
print("Lowest average subject:", lowest_subject)

print("Top performing student:", highest_student)
print("Top student average:", round(highest_average, 2))

print("\nGrade Distribution:")
print(grade_counts)
```
